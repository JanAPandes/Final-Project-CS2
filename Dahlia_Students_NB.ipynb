```python
ref = db.reference("students")
students = ref.get()

data_student = []

if students:
    if isinstance(students, dict):
        data_student.extend(students.values())
    elif isinstance(students, list):
        data_student.extend([item for item in students if item is not None and isinstance(item, dict)])

def get_valid_grade(prompt):
    while True:
        try:
            grade = float(input(prompt))
            if 1.00 <= grade <= 5.00:
                return grade
            else:
                print("Invalid grade. Grade must be between 1.00 and 5.00.")
        except ValueError:
            print("Invalid input. Please enter a numerical grade.")

while True:
    print("\n===== STUDENT DATABASE MENU =====")
    print("1. Display Students")
    print("2. Add Student")
    print("3. Update Student")
    print("4. Delete Student")
    print("5. Features")
    print("6. Exit")

    choice = input("Enter choice: ")

    # DISPLAY
    if choice == "1":
        print("\nStudent List:")
        if not data_student:
            print("No students found.")
        else:
            for s in data_student:
                print(f"ID: {s['id']}, Name: {s['name']}, Section: {s['section']}, Grades: {s['subjects']}")


    # ADD
    elif choice == "2":
        sid = input("Enter ID: ")
        if not sid:
            print("Student ID cannot be empty.")
            continue
        if ref.child(sid).get():
            print("Student already exists!")
            continue
        name = input("Enter name: ")

        section = input("Enter section: ")

        math2 = get_valid_grade("Enter Math 2 grade: ")
        math3 = get_valid_grade("Enter Math 3 grade: ")
        bio = get_valid_grade("Enter Biology 1 grade: ")
        chem = get_valid_grade("Enter Chemistry 1 grade: ")
        physics = get_valid_grade("Enter Physics 1 grade: ")
        cs = get_valid_grade("Enter Computer Science 2 grade: ")
        pehm = get_valid_grade("Enter PEHM2 grade: ")
        ve = get_valid_grade("Enter VE2 grade: ")
        ss = get_valid_grade("Enter Social Science 2 grade: ")
        eng = get_valid_grade("Enter English 2 grade: ")
        fili = get_valid_grade("Enter Filipino 2 grade: ")

        student = {
            "id": int(sid),
            "name": name,
            "section": section,
            "subjects": {
                "Math2": math2,
                "Math3": math3,
                "Biology1": bio,
                "Chem1": chem,
                "Physics1": physics,
                "Computer Science2": cs,
                "PEHM2": pehm,
                "VE2": ve,
                "Social Science2": ss,
                "English2": eng,
                "Filipino2": fili
            }
        }

        ref.child(sid).set(student)

        students = ref.get()
        data_student = []
        if students:
            if isinstance(students, dict):
                data_student.extend(students.values())
            elif isinstance(students, list):
                data_student.extend([item for item in students if item is not None and isinstance(item, dict)])
        print("Student added successfully!")

    # UPDATE
    elif choice == "3":
        sid = input("Enter ID of student to update: ")
        if not sid:
            print("Student ID cannot be empty.")
            continue
        student_ref = ref.child(sid)

        student_to_update = student_ref.get()

        if student_to_update:
            name = input("Enter new name: ")
            section = input("Enter new section: ")
            math2 = get_valid_grade("Enter Math 2 grade: ")
            math3 = get_valid_grade("Enter Math 3 grade: ")
            bio = get_valid_grade("Enter Biology 1 grade: ")
            chem = get_valid_grade("Enter Chemistry 1 grade: ")
            physics = get_valid_grade("Enter Physics 1 grade: ")
            cs = get_valid_grade("Enter Computer Science 2 grade: ")
            pehm = get_valid_grade("Enter PEHM2 grade: ")
            ve = get_valid_grade("Enter VE2 grade: ")
            ss = get_valid_grade("Enter Social Science 2 grade: ")
            eng = get_valid_grade("Enter English 2 grade: ")
            fili = get_valid_grade("Enter Filipino 2 grade: ")

            student_ref.update({
                "name": name,
                "section": section,
                  "subjects": {
                  "Math2": math2,
                  "Math3": math3,
                  "Biology1": bio,
                  "Chem1": chem,
                  "Physics1": physics,
                  "Computer Science2": cs,
                  "PEHM2": pehm,
                  "VE2": ve,
                  "Social Science2": ss,
                  "English2": eng,
                  "Filipino2": fili
                }
            })
            print("Student updated successfully!")

            students = ref.get()
            data_student = []
            if students:
                if isinstance(students, dict):
                    data_student.extend(students.values())
                elif isinstance(students, list):
                    data_student.extend([item for item in students if item is not None and isinstance(item, dict)])
        else:
            print("Student not found.")

    # DELETE
    elif choice == "4":
        sid = input("Enter ID to delete: ")
        if not sid:
            print("Student ID cannot be empty.")
            continue
        if ref.child(sid).get():
            ref.child(sid).delete()
            print("Student deleted successfully!")

            students = ref.get()
            data_student = []
            if students:
                if isinstance(students, dict):
                    data_student.extend(students.values())
                elif isinstance(students, list):
                    data_student.extend([item for item in students if item is not None and isinstance(item, dict)])
            else:
                data_student = []
        else:
            print("Student not found.")

    # FEATURES
    elif choice == "5":
      print("Features")
      while True:
        print("\n===== STUDENT DATABASE FEATURES =====")
        print("1. Compute the General Average of One Student.")
        print("2. List the Subjects Where the Student Scored 2.5 or Better.")
        print("3. List the Subjects Where the Student Scored 2.75 or Worse.")
        print("4. List the Students Who Have a Substandard General Average or Worse.")
        print("5. List the Students Who Have Qualified for a Director's List in Their General Average.")
        print("6. Back to Main Menu.")
        choice = input("Enter choice: ")

        match choice:
          case "1":
            id = input("Enter student ID: ")
            if not id:
                print("Student ID cannot be empty.")
                continue
            student_info = ref.child(id).get()

            if not student_info:
                print("Student not found.")
                continue

            sum_grades = 0
            grades_count = 0
            for subject, grade in student_info['subjects'].items():
                sum_grades += grade
                grades_count += 1

            gen_avg = sum_grades / grades_count
            print(f"The general average of {student_info['name']} is {gen_avg:.2f}!")

          case "2":
            id = input("Enter student ID: ")
            if not id:
                print("Student ID cannot be empty.")
                continue
            student_info = ref.child(id).get()

            if not student_info:
                print("Student not found.")
                continue

            print(f"Subjects where {student_info['name']} scored 2.5 or better:")
            found_subjects = False
            for subject, grade in student_info['subjects'].items():
              if grade <= 2.5:
                print(f"  {subject} - {grade}")
                found_subjects = True
            if not found_subjects:
                print("  No subjects found.")

          case "3":
            id = input("Enter student ID: ")
            if not id:
                print("Student ID cannot be empty.")
                continue
            student_info = ref.child(id).get()

            if not student_info:
                print("Student not found.")
                continue

            print(f"Subjects where {student_info['name']} scored 2.75 or worse:")
            found_subjects = False
            for subject, grade in student_info['subjects'].items():
              if grade >= 2.75:
                print(f"  {subject} - {grade}")
                found_subjects = True
            if not found_subjects:
                print("No subjects found.")

          case "4":
            print("Students with a substandard General Average (2.75 or worse):")
            all_students = ref.get()
            found_students = False
            if all_students:
                students_list = []
                if isinstance(all_students, dict):
                    students_list = list(all_students.values())
                elif isinstance(all_students, list):
                    students_list = [s for s in all_students if s is not None and isinstance(s, dict)]

                for student_info in students_list:
                    sum_grades = 0
                    grades_count = 0
                    for grade_val in student_info['subjects'].values():
                        sum_grades += grade_val
                        grades_count += 1
                    if grades_count > 0:
                        gen_avg = sum_grades / grades_count
                        if gen_avg >= 2.75:
                            print(f"  {student_info['name']} - {gen_avg:.2f}")
                            found_students = True
            if not found_students:
                print("No students found.")

          case "5":
            print("Students who have qualified for Director's List (General Average 1.50 or better):")
            all_students = ref.get()
            found_students = False
            if all_students:
                students_list = []
                if isinstance(all_students, dict):
                    students_list = list(all_students.values())
                elif isinstance(all_students, list):
                    students_list = [s for s in all_students if s is not None and isinstance(s, dict)]

                for student_info in students_list:
                    sum_grades = 0
                    grades_count = 0
                    for grade_val in student_info['subjects'].values():
                        sum_grades += grade_val
                        grades_count += 1
                    if grades_count > 0:
                        gen_avg = sum_grades / grades_count
                        if gen_avg <= 1.50:
                            print(f"  {student_info['name']} - {gen_avg:.2f}")
                            found_students = True
            if not found_students:
                print("No students found.")

          case "6":
            break

          case _:
            print("Invalid choice. Try again.")

    # EXIT
    elif choice == "6":
        print("Exiting program...")
        break

    else:
        print("Invalid choice. Try again.")
```
