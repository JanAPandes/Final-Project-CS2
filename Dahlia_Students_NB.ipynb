```python
# FEATURE 1

id = input("Enter student ID: ")

ref = db.reference("students")
student = ref.child(id).get()

sum = 0

data_student = []
grades = []
for s in student:
  data_student.append(student[s])

for s in data_student[3]:
  grades.append(data_student[3])

for s in grades[0]:
  sum += grades[0][s]

gen_avg = sum / len(grades[0])
print(f"The general average of {data_student[1]} is {gen_avg:.2f}!")
```

```python
ref = db.reference("students")
student = ref.get()
student.remove(None)

data_student = []
counter = 0
for s in student:
  data_student.append(student[counter])
  counter += 1
counter = 0

while True:
    print("\n===== STUDENT DATABASE MENU =====")
    print("1. Display Students")
    print("2. Add Student")
    print("3. Update Student")
    print("4. Delete Student")
    print("5. Exit")

    choice = input("Enter choice: ")

    # DISPLAY
    if choice == "1":
        print("\nStudent List:")
        for s in data_student:
            print(f"ID: {s['id']}, Name: {s['name']}, Section: {s['section']}, Grades: {s['subjects']}")


    # ADD
    elif choice == "2":
        sid = input("Enter ID: ")
        if sid in data_student:
            print("Student already exists!")
            continue
        name = input("Enter name: ")
        section = input("Enter section: ")
        math2 = float(input("Enter Math 2 grade: "))
        math3 = float(input("Enter Math 3 grade: "))
        bio = float(input("Enter Biology 1 grade: "))
        chem = float(input("Enter Chemistry 1 grade: "))
        physics = float(input("Enter Physics 1 grade: "))
        cs = float(input("Enter Computer Science 2 grade: "))
        pehm = float(input("Enter PEHM2 grade: "))
        ve = float(input("Enter VE2 grade: "))
        ss = float(input("Enter Social Science 2 grade: "))
        eng = float(input("Enter English 2 grade: "))
        fili = float(input("Enter Filipino 2 grade: "))

        student = {
            "id": int(sid),
            "name": name,
            "section": section,
            "subjects": {
                "Math2": math2,
                "Math3": math3,
                "Biology1": bio,
                "Chem1": chem,
                "Physics1": physics,
                "Computer Science2": cs,
                "PEHM2": pehm,
                "VE2": ve,
                "Social Science2": ss,
                "English2": eng,
                "Filipino2": fili
            }
        }

        ref.child(sid).set(student)
        print("Student added successfully!")

    # UPDATE
    elif choice == "3":
        sid = input("Enter ID of student to update: ")
        student_ref = ref.child(sid)

        student = student_ref.get()

        if student:
            name = input("Enter new name: ")
            math2 = float(input("Enter Math 2 grade: "))
            math3 = float(input("Enter Math 3 grade: "))
            bio = float(input("Enter Biology 1 grade: "))
            chem = float(input("Enter Chemistry 1 grade: "))
            physics = float(input("Enter Physics 1 grade: "))
            cs = float(input("Enter Computer Science 2 grade: "))
            pehm = float(input("Enter PEHM2 grade: "))
            ve = float(input("Enter VE2 grade: "))
            ss = float(input("Enter Social Science 2 grade: "))
            eng = float(input("Enter English 2 grade: "))
            fili = float(input("Enter Filipino 2 grade: "))

            student_ref.update({
                "name": name,
                  "subjects": {
                  "Math2": math2,
                  "Math3": math3,
                  "Biology1": bio,
                  "Chem1": chem,
                  "Physics1": physics,
                  "Computer Science2": cs,
                  "PEHM2": pehm,
                  "VE2": ve,
                  "Social Science2": ss,
                  "English2": eng,
                  "Filipino2": fili
                }
            })

            print("Student updated successfully!")
        else:
            print("Student not found.")

    # DELETE
    elif choice == "4":
        sid = input("Enter ID to delete: ")

        if ref.child(sid).get():
            ref.child(sid).delete()
            print("Student deleted successfully!")
        else:
            print("Student not found.")

    # EXIT
    elif choice == "5":
        print("Exiting program...")
        break

    else:
        print("Invalid choice. Try again.")

"""
Loop → keeps menu running

Conditionals (if-elif-else) → handles choices
CRUD operations:
Create → set()
Read → get()
Update → update()
Delete → delete()
Dictionary (JSON) → data structure
Firebase interaction
"""
```
