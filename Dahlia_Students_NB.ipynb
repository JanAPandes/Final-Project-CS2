```python
# FEATURE 1

id = input("Enter student ID: ")

ref = db.reference("students")
student = ref.child(id).get()

sum = 0

data_student = []
grades = []
for s in student:
  data_student.append(student[s])

for s in data_student[3]:
  grades.append(data_student[3])

for s in grades[0]:
  sum += grades[0][s]

gen_avg = sum / len(grades[0])
print(f"The general average of {data_student[1]} is {gen_avg:.2f}!")
```
